```python
import compas
from compas.geometry import Box
from compas_notebook import Viewer
from compas_notebook.config import Config

print(f"COMPAS {compas.__version__}")
```

## 2. Create a box and a mesh
The dimensions are in metres. Change the three numbers, then rerun this cell and the viewer cell.


```python
box = Box(xsize=2.0, ysize=1.0, zsize=1.5)
mesh = box.to_mesh()

print(mesh)
print(f"Volume: {box.volume:g} m³")
print(f"Surface area: {box.area:g} m²")
```

## 3. Add the mesh to a COMPAS scene
`Viewer`, `viewer.scene.add(mesh)` and `viewer.show()` are the same APIs used in the official COMPAS notebooks. The mesh is interactive: orbit, pan and zoom to inspect its geometry.


```python
config = Config()
config.view.width = 700
config.view.height = 420
config.view.camera.position = [4, -6, 3]
config.view.camera.target = [0, 0, 0]

viewer = Viewer(config=config)
viewer.scene.add(mesh, color="#5552bb", show_vertices=True, show_edges=True)
viewer.show()
```
