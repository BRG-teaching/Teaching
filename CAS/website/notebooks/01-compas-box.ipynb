## 1. Load COMPAS
Run this cell once after starting or restarting the kernel. `piplite` installs packages in browser Python. In desktop Jupyter or Google Colab, this cell installs missing packages with pip instead.


```python
import sys
if sys.platform == "emscripten":
    import piplite
    await piplite.install(["compas==2.15.1", "matplotlib"])
else:
    # Desktop Jupyter / Google Colab: install missing packages in this kernel.
    import importlib.util
    import subprocess
    if any(importlib.util.find_spec(name) is None for name in ["compas", "matplotlib"]):
        subprocess.check_call([sys.executable, "-m", "pip", "install", "compas==2.15.1", "matplotlib"])

import compas
from compas.geometry import Box
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d.art3d import Poly3DCollection

print(f"COMPAS {compas.__version__} is ready.")
```

## 2. Create a box
The dimensions are in metres. Change the three numbers below, then rerun this cell and the drawing cell.


```python
xsize = 2.0
ysize = 1.0
zsize = 1.5

box = Box(xsize=xsize, ysize=ysize, zsize=zsize)
vertices, faces = box.to_vertices_and_faces()

print(f"Dimensions: {xsize:g} × {ysize:g} × {zsize:g} m")
print(f"Volume: {box.volume:g} m³")
print(f"Surface area: {box.area:g} m²")
print(f"Geometry: {len(vertices)} vertices, {len(faces)} faces")
```

## 3. Draw the geometry
Each face is a polygon made from the COMPAS box's vertices. Matplotlib displays the result directly in the notebook.


```python
fig = plt.figure(figsize=(6, 4.5))
ax = fig.add_subplot(111, projection="3d")
polygons = [[vertices[index] for index in face] for face in faces]
ax.add_collection3d(Poly3DCollection(
    polygons, facecolors="#5552bb", edgecolors="#141b3d",
    linewidths=1.2, alpha=0.35,
))

# Equal scale on all axes makes the proportions of the box visible.
limit = max(xsize, ysize, zsize) * 0.65
ax.set(xlim=(-limit, limit), ylim=(-limit, limit), zlim=(-limit, limit),
       xlabel="x [m]", ylabel="y [m]", zlabel="z [m]",
       title=f"COMPAS box · volume {box.volume:g} m³")
ax.set_box_aspect((1, 1, 1))
ax.view_init(elev=22, azim=35)
plt.tight_layout()
plt.show()
```
